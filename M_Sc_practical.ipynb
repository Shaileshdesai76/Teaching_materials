```python
%%writefile analysis.R
# Before moving forward, let's modify the data for plotting.
# We have more than 1000 samples, so we need to organise them
# geographically, linguistically, or according to our research question.

library(readr)
library(ggplot2)
library(dplyr)

# STEP 1: Read PCA eigenvectors

Data1 <- read_table(
  "plink.eigenvec",
  col_names = c(
    "POPULATION",
    "INDIVIDUAL",
    paste0("PC", 1:10)
  ),
  show_col_types = FALSE
)

# STEP 2: Read population classification file

Data2 <- read_delim(
  "pop_PCA_corrobo",
  delim = "\t",
  col_names = TRUE,
  show_col_types = FALSE
)

head(Data1)
head(Data2)


# STEP 3: Check whether all individuals are covered

unmatched <- Data1[
  !Data1$INDIVIDUAL %in% Data2$`INDIVIDUAL ID`,
]

head(unmatched)

nrow(unmatched)  # Ideally, this should be zero


# STEP 4: Merge the datasets

Data <- merge(
  Data1,
  Data2,
  by.x = c("POPULATION", "INDIVIDUAL"),
  by.y = c("POPULATION", "INDIVIDUAL ID")
)

head(Data)
dim(Data)


# STEP 5: Basic PCA plot

plot(
  Data$PC1,
  Data$PC2,
  main = "Principal Component Analysis",
  xlab = "PCA 1",
  ylab = "PCA 2"
)


# STEP 6: PCA plot using ggplot2

p <- ggplot(Data, aes(x = PC1, y = PC2)) +

  geom_point(
    aes(colour = factor(GROUP)),
    size = 1.5,
    alpha = 0.8
  ) +

  labs(
    colour = "Geographical Group",
    x = "PC1",
    y = "PC2"
  ) +

  theme_classic()

# Display plot
print(p)


# STEP 7: Save PCA plots

ggsave(
  "PCA_plot_colour.tiff",
  plot = p,
  width = 8,
  height = 6,
  units = "in",
  dpi = 300,
  compression = "lzw"
)

ggsave(
  "PCA_plot_colour.pdf",
  plot = p,
  width = 8,
  height = 6,
  units = "in"
)

ggsave(
  "PCA_plot_colour.png",
  plot = p,
  width = 8,
  height = 6,
  units = "in",
  dpi = 300
)
```
